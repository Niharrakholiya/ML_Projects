```python
%pip install pandas numpy scikit-learn

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
import re

# Load data function
def load_fasttext_data(filepath):
    texts = []
    labels = []
    with open(filepath, 'r', encoding='utf-8') as f:
        for line in f:
            parts = line.strip().split(' ', 1)
            label = 'Negative' if parts[0] == '__label__1' else 'Positive'
            text = parts[1]
            texts.append(text)
            labels.append(label)
    return texts, labels

# Load training and test data
train_texts, train_labels = load_fasttext_data('./train.ft.txt')
test_texts, test_labels = load_fasttext_data('./test.ft.txt')

# Preprocessing
def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    return text

# Preprocess texts
train_processed = [preprocess_text(text) for text in train_texts]
test_processed = [preprocess_text(text) for text in test_texts]

# Vectorization
vectorizer = TfidfVectorizer(max_features=5000)
X_train = vectorizer.fit_transform(train_processed)
X_test = vectorizer.transform(test_processed)

# Naive Bayes
nb_model = MultinomialNB()
nb_model.fit(X_train, train_labels)
nb_pred = nb_model.predict(X_test)

print("Naïve Bayes Results:")
print(classification_report(test_labels, nb_pred))

# Logistic Regression
lr_model = LogisticRegression(max_iter=1000)
lr_model.fit(X_train, train_labels)
lr_pred = lr_model.predict(X_test)

print("\nLogistic Regression Results:")
print(classification_report(test_labels, lr_pred))

# Prediction function
def predict_sentiment(review, model='naive_bayes'):
    processed_review = preprocess_text(review)
    vectorized_review = vectorizer.transform([processed_review])
    
    if model == 'naive_bayes':
        return nb_model.predict(vectorized_review)[0]
    return lr_model.predict(vectorized_review)[0]

# Test predictions
test_reviews = [
    'This product is fantastic',
    'I really hate this product',
    'Nothing special about it'
]

print("\nTest Predictions:")
for review in test_reviews:
    nb_result = predict_sentiment(review, 'naive_bayes')
    lr_result = predict_sentiment(review, 'logistic_regression')
    print(f"Review: {review}")
    print(f"Naïve Bayes: {nb_result}")
    print(f"Logistic Regression: {lr_result}\n")
```
